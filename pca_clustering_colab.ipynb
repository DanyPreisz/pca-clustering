# PCA + Clustering

Proyecto de reducción de dimensionalidad + clustering listo para **Google Colab**.

**Pregunta:** ¿podemos reducir dimensiones y segmentar datos de alta dimensionalidad?

**Método:** PCA + K-Means.

**Stack:** pandas, seaborn, scikit-learn

Dataset de ejemplo (Iris o simulado).

## 0. Setup

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
print("OK")
```

## 1. Cargar datos

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = iris.target_names[iris.target]
print(df.shape)
df.head()
```

## 2. EDA

```python
print(df.describe().T)
```

```python
sns.pairplot(df, hue="species", height=2)
plt.suptitle("Pairplot Iris", y=1.02)
plt.show()
```

## 3. Preprocesamiento

```python
X = df.drop(columns=["species"])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled.shape)
```

## 4. PCA

```python
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_scaled)
print("Varianza explicada:", pca.explained_variance_ratio_.round(3))
print("Total:", pca.explained_variance_ratio_.sum().round(3))
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=df["species"])
plt.title("PCA 2D - Iris")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

## 5. Clustering sobre PCA

```python
k_final = 3
km = KMeans(n_clusters=k_final, random_state=RANDOM_STATE, n_init=10)
labels = km.fit_predict(X_pca)
print("Silhouette:", silhouette_score(X_pca, labels).round(3))
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], hue=labels, palette="tab10")
plt.title("Clusters sobre PCA")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

## 6. Comparación con especie real

```python
df["cluster"] = labels
pd.crosstab(df["species"], df["cluster"])
```

## 7. Conclusiones

- PCA reduce a 2D preservando la mayor parte de la varianza.
- K-Means sobre el espacio reducido recupera bien las especies.
- Técnica útil cuando hay muchas features.
